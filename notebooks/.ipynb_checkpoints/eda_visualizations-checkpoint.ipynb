```python
import pandas as pd
import numpy as np
```

```python
imports_df = pd.read_csv('../outputs/census_imports.csv')
exports_df = pd.read_csv('../outputs/census_exports.csv')
gdp_df = pd.read_csv('../outputs/gdp_states.csv')
```

```python
gdp_overall = gdp_df[gdp_df['NAICS'] == '..0'][['STATE/REGION','YEAR','GDP']].rename(columns={'GDP':'OVERALL_GDP'})
gdp_df = gdp_df.merge(gdp_overall, on=['STATE/REGION','YEAR'])
```

### Grouping Columns

```python
bea_regions = {
    # New England
    'CT': 'New England', 'ME': 'New England', 'MA': 'New England', 
    'NH': 'New England', 'RI': 'New England', 'VT': 'New England',
    # Mideast
    'DE': 'Mideast', 'DC': 'Mideast', 'MD': 'Mideast', 
    'NJ': 'Mideast', 'NY': 'Mideast', 'PA': 'Mideast',
    # Great Lakes
    'IL': 'Great Lakes', 'IN': 'Great Lakes', 'MI': 'Great Lakes', 
    'OH': 'Great Lakes', 'WI': 'Great Lakes',
    # Plains
    'IA': 'Plains', 'KS': 'Plains', 'MN': 'Plains', 'MO': 'Plains', 
    'NE': 'Plains', 'ND': 'Plains', 'SD': 'Plains',
    # Southeast
    'AL': 'Southeast', 'AR': 'Southeast', 'FL': 'Southeast', 'GA': 'Southeast', 
    'KY': 'Southeast', 'LA': 'Southeast', 'MS': 'Southeast', 'NC': 'Southeast', 
    'SC': 'Southeast', 'TN': 'Southeast', 'VA': 'Southeast', 'WV': 'Southeast',
    # Southwest
    'AZ': 'Southwest', 'NM': 'Southwest', 'OK': 'Southwest', 'TX': 'Southwest',
    # Rocky Mountain
    'CO': 'Rocky Mountain', 'ID': 'Rocky Mountain', 'MT': 'Rocky Mountain', 
    'UT': 'Rocky Mountain', 'WY': 'Rocky Mountain',
    # Far West
    'AK': 'Far West', 'CA': 'Far West', 'HI': 'Far West', 
    'NV': 'Far West', 'OR': 'Far West', 'WA': 'Far West'
}

gdp_df['BEA_REGION'] = gdp_df['STATE/REGION'].map(bea_regions)
```

```python
era_map = {
    2010: 'Late-Obama Era',
    2011: 'Late-Obama Era',
    2012: 'Late-Obama Era',
    2013: 'Late-Obama Era',
    2014: 'Late-Obama Era',
    2015: 'Late-Obama Era',
    2016: 'Late-Obama Era',
    2017: 'Pre-COVID Era',
    2018: 'Pre-COVID Era',
    2019: 'Pre-COVID Era',
    2020: 'Post-COVID Era',
    2021: 'Post-COVID Era',
    2022: 'Post-COVID Era',
    2023: 'Post-COVID Era',
    2024: 'Post-COVID Era',
    2025: 'Post-COVID Era'
}

gdp_df['TIME_PERIOD'] = gdp_df['YEAR'].map(era_map)
```

## Growth Contribution

```python
gdp_df.sort_values(by=['STATE/REGION','NAICS','YEAR'])

gdp_df['YoY CHANGE'] = gdp_df.groupby(['STATE/REGION','NAICS'])['GDP'].pct_change() * 100
gdp_df['GDP_SHARE'] = gdp_df['GDP'] / gdp_df['OVERALL_GDP']
gdp_df['GDP_SHARE_PY'] = gdp_df.groupby(['STATE/REGION','NAICS'])['GDP_SHARE'].shift(1)
gdp_df['GROWTH_CONTRIBUTION'] = gdp_df['GDP_SHARE_PY'] * gdp_df['YoY CHANGE']
```

### Sector Group GDP Growth

```python
gdp_gsector = (gdp_df[gdp_df['LEVEL'] == 0]
                  .drop(columns=['NAICS'])
                  .rename(columns={'NAICS_LDESC':'NAICS'}))
```

```python
gdp_gsector
```

```python
gdp_gsector_growth = gdp_gsector.groupby(['STATE/REGION','NAICS'])['GROWTH_CONTRIBUTION'].sum().reset_index()
gdp_gsector_growth['RANK'] = gdp_gsector_growth.groupby('STATE/REGION')['GROWTH_CONTRIBUTION'].rank(ascending=False,method='min')

# Getting average rank across all states
(gdp_gsector_growth
    .groupby('NAICS')['RANK'].mean().reset_index()
    .sort_values(by=['RANK'])
    .rename(columns={'RANK':'AVG_RANK'}))
```

```python
gdp_gsector_reg_growth = gdp_gsector.groupby(['BEA_REGION','NAICS'])['GROWTH_CONTRIBUTION'].sum().reset_index()
gdp_gsector_reg_growth['RANK'] = gdp_gsector_reg_growth.groupby('BEA_REGION')['GROWTH_CONTRIBUTION'].rank(ascending=False,method='min')

# Getting top 5 ranked growth for each region
(gdp_gsector_reg_growth
    .loc[gdp_gsector_reg_growth['RANK'] <= 5]
    .sort_values(by=['BEA_REGION','RANK']))
```

```python
# Getting top 15 growth for all regions & sector groups
gdp_gsector_reg_growth.sort_values(by=['GROWTH_CONTRIBUTION'],ascending=False).head(15)
```

### Sector GDP Growth

```python
gdp_sector = (gdp_df[gdp_df['LEVEL'] == 2]
                  .drop(columns=['NAICS'])
                  .rename(columns={'NAICS_LDESC':'NAICS'}))
```

```python
gdp_sector
```

```python
gdp_sector_growth = gdp_sector.groupby(['STATE/REGION','NAICS'])['GROWTH_CONTRIBUTION'].sum().reset_index()
gdp_sector_growth['RANK'] = gdp_sector_growth.groupby('STATE/REGION')['GROWTH_CONTRIBUTION'].rank(ascending=False,method='min')

# Getting average rank across all states
(gdp_sector_growth
    .groupby('NAICS')['RANK'].mean().reset_index()
    .sort_values(by=['RANK'])
    .rename(columns={'RANK':'AVG_RANK'}))
```

```python
gdp_sector_time_growth = gdp_sector.groupby(['TIME_PERIOD','NAICS'])['GROWTH_CONTRIBUTION'].sum().reset_index()
gdp_sector_time_growth['RANK'] = gdp_sector_time_growth.groupby('TIME_PERIOD')['GROWTH_CONTRIBUTION'].rank(ascending=False,method='min')

# Getting top 5 growth sectors for each time period
(gdp_sector_time_growth
    .loc[gdp_sector_time_growth['RANK'] <= 5]
    .sort_values(by=['TIME_PERIOD','RANK']))
```

```python
gdp_sector_reg_growth = gdp_sector.groupby(['BEA_REGION','NAICS'])['GROWTH_CONTRIBUTION'].sum().reset_index()
gdp_sector_reg_growth['RANK'] = gdp_sector_reg_growth.groupby('BEA_REGION')['GROWTH_CONTRIBUTION'].rank(ascending=False,method='min')

# Getting top 5 ranked growth for each region
(gdp_sector_reg_growth
    .loc[gdp_sector_reg_growth['RANK'] <= 5]
    .sort_values(by=['BEA_REGION','RANK']))
```

```python
# Getting top 10 growth for all regions & sectors
gdp_sector_reg_growth.sort_values(by=['GROWTH_CONTRIBUTION'],ascending=False).head(10)
```

### Subsector GDP Growth

```python
gdp_subsector = (gdp_df[gdp_df['LEVEL'] == 3]
                  .drop(columns=['NAICS'])
                  .rename(columns={'NAICS_LDESC':'NAICS'}))
```

```python
gdp_subsector
```

```python
gdp_subsector_growth = gdp_subsector.groupby(['STATE/REGION','NAICS'])['GROWTH_CONTRIBUTION'].sum().reset_index()
gdp_subsector_growth['RANK'] = gdp_subsector_growth.groupby('STATE/REGION')['GROWTH_CONTRIBUTION'].rank(ascending=False,method='min')

# Getting average rank across all states
(gdp_subsector_growth
    .groupby('NAICS')['RANK'].mean().reset_index()
    .sort_values(by=['RANK'])
    .rename(columns={'RANK':'AVG_RANK'}))
```

```python
# Getting top 5 ranked growth for each region
(gdp_subsector_reg_growth
    .loc[gdp_subsector_reg_growth['RANK'] <= 5]
    .sort_values(by=['BEA_REGION','RANK']))
```

```python
# Getting top 15 growth for all regions & sectors
gdp_subsector_reg_growth.sort_values(by=['GROWTH_CONTRIBUTION'],ascending=False).head(15)
```

## Compound Annual Growth Rate (CAGR)

```python
gdp_df['GDP_CAPITA'] = gdp_df['GDP'] / gdp_df['EST_POP']
```

```python
gdp_2010 = (gdp_df
                .loc[gdp_df['YEAR'] == 2010][[
                    'REGION_NAME',
                    'STATE/REGION',
                    'NAICS',
                    'NAICS_LDESC',
                    'LEVEL',
                    'GDP',
                    'GDP_CAPITA',
                    'BEA_REGION']])
gdp_2024 = (gdp_df
                .loc[gdp_df['YEAR'] == 2024][[
                    'REGION_NAME',
                    'STATE/REGION',
                    'NAICS',
                    'NAICS_LDESC',
                    'LEVEL',
                    'GDP',
                    'GDP_CAPITA',
                    'BEA_REGION']])
```

```python
gdp_CAGR = gdp_2010.merge(gdp_2024, on=['REGION_NAME','STATE/REGION','NAICS','NAICS_LDESC','LEVEL','BEA_REGION'])
gdp_CAGR['CAGR'] = ((gdp_CAGR['GDP_y'] / gdp_CAGR['GDP_x']) ** (1/14) - 1) * 100
gdp_CAGR['CAGR_CAPITA'] = ((gdp_CAGR['GDP_CAPITA_y'] / gdp_CAGR['GDP_CAPITA_x']) ** (1/14) - 1) * 100
```

### Sector Group

```python
gdp_gsector_CAGR = gdp_CAGR.loc[gdp_CAGR['LEVEL'] == 0]
```

```python
(gdp_gsector_CAGR
    .groupby('NAICS_LDESC')['CAGR'].mean().reset_index()
    .sort_values(by='CAGR',ascending=False)).head(5)
```

### Sector

```python
gdp_sector_CAGR = gdp_CAGR.loc[gdp_CAGR['LEVEL'] == 2]
```

```python
(gdp_sector_CAGR
    .groupby('NAICS_LDESC')['CAGR'].mean().reset_index()
    .sort_values(by='CAGR',ascending=False)
    .rename(columns={'CAGR':'AVG_CAGR'})).head(5)
```

```python
(gdp_sector_CAGR
    .groupby(['BEA_REGION','NAICS_LDESC'])['CAGR_CAPITA'].mean().reset_index()
    .sort_values(by='CAGR_CAPITA',ascending=False)
    .rename(columns={'CAGR_CAPITA':'AVG_CAGR_CAPITA'})).head(10)
```

```python
(gdp_sector_CAGR
    .groupby(['STATE/REGION','NAICS_LDESC'])['CAGR_CAPITA'].mean().reset_index()
    .sort_values(by='CAGR_CAPITA',ascending=False)).head(30)
```
